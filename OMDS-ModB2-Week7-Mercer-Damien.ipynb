# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
"""To draw a viewers attention to the last three months of data various preattentive attributes could be used. One option would be to use a different color that would contrast the last three months of data from the earlier months. Another option would be making the line thicker for those last three months to make that portion stand out. A third option could be to add larger marks on the last three months of data. Lastly, the last three months could be highlighted with a shaded background to make the eyes of the viewer be drawn to that section."""
```

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

#Categories
concerns=[
    'Engine Power is less than expected',
    'Tires make excessive noise while driving',
    'Engine makes abnormal/excessive noise',
    'Seat material concerns',
    'Excessive wind noise',
    'Hesitation or delay when shifting',
    'Bluetooth system has poor sound quality',
    'Steering system/wheel has too much play',
    'Bluetooth system is difficult to use',
    'Front seat audio/entertainment/navigation controls']

#Values for concerns
values=[12.9,12.3,11.6,11.6,11.0,10.3,10.0,8.8,8.6,8.2]

#Colors for the concerns
colors=['#e4aaaa',
        '#ff4d4d',
        '#ff4d4d',
        '#e4aaaa',
        '#ff4d4d',
        '#e4aaaa',
        '#e4aaaa',
        '#bfbfbf',
        '#bfbfbf',
        '#bfbfbf',]

#Figure
fig, ax = plt.subplots(figsize=(12,7))

#Bars for chart
bars = ax.barh(concerns, values, color=colors, height=0.7)

#order of concerns
ax.invert_yaxis()

#Values in bars
for bar, value in zip(bars, values):
    ax.text(
        value - .25,
        bar.get_y() + bar.get_height() / 2,
        f'{value:.1f}',
        va='center',
        ha='right',
        color='white',
        fontsize=9)

#Title / Headings
ax.set_title(
    'Top 10 design concerns',
    loc='left',
    fontsize=14,
    color='dimgray',
    pad=15)
    
ax.text(
    7.5,
    -.95,
    'concerns per 1000',
    fontsize=9,
    color='gray')
    
#Remove 
ax.set_xticks([])
for spine in ['top', 'right', 'bottom', 'left']:
    ax.spines[spine].set_visible(False)
ax.tick_params(axis='y',length=0)

#Labels
ax.tick_params(axis='y', labelsize=9, colors='dimgray')

#Text
ax.set_xlim(0,21)
ax.text(13.5,1,
        'Comments indicate that\n'
        'noisy tire issues are\n'
        'most apparent in the rain.',
        fontsize=9,
        color='dimgray',va='center'
        )

ax.text(13.5,2.8,
        'Complaints about engine\n'
        'noise commonly cited\n'
        'after the car had not\n'
        'been driven for a while.',
        fontsize=9,
        color='dimgray',va='center')

ax.text(13.5,4.5,
        'Excessive wind noise is\n'
        'noted primarily freeway\n'
        'driving at high speeds.',
        fontsize=9,
        color='dimgray',va='center')

plt.tight_layout()
plt.show()
```
