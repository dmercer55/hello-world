# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
#Advantages and Disadvantages of Chart Presentations

"""
The side by side graph makes it easier to review the pattern of visitors for each week separatel without the lines overlapping. This can make each week's individual trend easier to read. However, a disadvantage is that comparing the same day across the two weeks is more challenging because you have to look back and forth between the graphs.

The combined graph makes is easier to compare the two weeks because both lines use the same axes. For example, the graph clearly shows hat the first week reaches its highest point at roughly 130 visitors on day 2, while the second week reaches its highest point at about 120 visitors on day 4. The disadvantage is that multiple lines on the same graph become harder to read, especially if more weeks are added. A legend would be helpful to include so that each week could be identified.

Overall, I would selected the combined graph because it makes differences and changes between the two weeks easier to compare.
"""
```

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
#Figures 8.17 - Retail Price Over Time

years = [2019,2020,2021,2022,2023,2024,2025]

#Product Prices
product_a = [390,420,420,390,300,270,255]
product_b = [360,400,410,380,285,250,240]

#Products C, D, and E with years
years_c = [2021,2022,2023,2024,2025]
product_c = [100,180,260,225,180]

years_d = [2022,2023,2024,2025]
product_d = [150,200,230,210]

years_e = [2024, 2025]
product_e = [100, 200]

#Create Figure
fig, ax = plt.subplots(figsize=(10,6))

#Products A and B

ax.plot(years, product_a, color='#404040', linewidth=2)
ax.plot(years, product_b, color='#595959', linewidth=2)

#Products C, D, and E

ax.plot(years_c, product_c, color='#1f4e79', linewidth=2)
ax.plot(years_d, product_d, color='#a6a6a6', linewidth=2)
ax.plot(years_e, product_e, color='#1f4e79', linewidth=2)

#Labels

ax.text(2018.8, 390, 'A', fontsize=10, color='dimgray')
ax.text(2018.8, 360, 'B', fontsize=10, color='dimgray')
ax.text(2020.7, 100, 'C', fontsize=10, color='dimgray')
ax.text(2021.7, 150, 'D', fontsize=10, color='dimgray')
ax.text(2023.7, 100, 'E', fontsize=10, color='dimgray')

#Title
ax.text(2018.5, 525, 'Retail Price Over Time', fontsize=10, color='dimgray')

#Y Axis
ax.set_ylabel('Average price', fontsize=9, color='dimgray')
ax.set_ylim(0, 500)
ax.set_yticks([0, 100, 200, 300, 400, 500])
ax.set_yticklabels(['$0', '$100', '$200', '$300', '$400', '$500'])

#X Axis

ax.set_xlabel('Year', fontsize=9, color='dimgray')
ax.set_xlim(2018.5, 2025.5)
ax.set_xticks(years)

#Format Chart
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#d9d9d9')
ax.spines['bottom'].set_color('#d9d9d9')

ax.grid(False)

ax.tick_params(axis='both', length=0, color='gray', labelsize=8)

plt.tight_layout()
plt.show()
```
